```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
import tensorflow as tf
from tensorflow import keras
import os
```

```python
# Import required libraries for downloading and extracting
import urllib.request
import zipfile
import os

# Dataset URL
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00240/UCI%20HAR%20Dataset.zip"

# File name to save
zip_path = "har_dataset.zip"

# Download the dataset
urllib.request.urlretrieve(url, zip_path)

print("Download complete.")

# Extract the dataset
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall("har_data")

print("Extraction complete.")
```

```python
X_train=pd.read_csv("har_data/UCI HAR Dataset/train/X_train.txt", delim_whitespace=True, header=None)
```

```python
y_train=pd.read_csv("har_data/UCI HAR Dataset/train/y_train.txt", delim_whitespace=True, header=None).iloc[:,0] - 1 #to make the labels start from 0 instead of 1
X_test=pd.read_csv("har_data/UCI HAR Dataset/test/X_test.txt", delim_whitespace=True, header=None)
y_test=pd.read_csv("har_data/UCI HAR Dataset/test/y_test.txt", delim_whitespace=True, header=None).iloc[:,0] - 1 #to make the labels start from 0 instead of 1
```

```python
print(X_train.isna().sum())
print(y_train.isna().sum())
print(X_test.isna().sum())
print(y_test.isna().sum())
```

```python
X_train
```

```python
y_train
```

```python
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
X_train=X_train_scaled[:,:20] #replaces the feature selection process for session purpose
X_test=X_test_scaled[:,:20] #replaces the feature selection process for session purpose
```

```python
y_train.shape
```

```python
model=keras.Sequential([
    keras.layers.Dense(256,activation="relu", input_shape=(20,)),
    keras.layers.Dense(128,activation="relu"),
    keras.layers.Dense(64,activation="relu"),
    keras.layers.Dense(16,activation="relu"),
    keras.layers.Dense(8,activation="relu"),
    keras.layers.Dense(6,activation="softmax")
])
```

```python
model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
```

```python
model.summary()
```

```python
model.fit(X_train, y_train, epochs=50, validation_data=(X_test, y_test))
```

```python
converter=tf.lite.TFLiteConverter.from_keras_model(model)
tflite_model=converter.convert()

with open("model.tflite", "wb") as f:
    f.write(tflite_model)
```

```python
converter=tf.lite.TFLiteConverter.from_keras_model(model)
converter.optimizations=[tf.lite.Optimize.DEFAULT]
tflite_model_quant=converter.convert()
```

```python
with open("model_quant.tflite", "wb") as f:
    f.write(tflite_model_quant)
```

```python
print("original model size:", os.path.getsize("model.tflite")/1024, "Kbytes")
print("quantized model size:", os.path.getsize("model_quant.tflite")/1024, "Kbytes")
```

```python
interpreter=tf.lite.Interpreter(model_path="model_quant.tflite")
interpreter.allocate_tensors()
```

```python
input_details=interpreter.get_input_details()
output_details=interpreter.get_output_details()
```

```python
sample=np.array([X_test[0]], dtype=np.float32)
interpreter.set_tensor(input_details[0]['index'], sample)
interpreter.invoke()
```

```python
output=interpreter.get_tensor(output_details[0]['index'])
```

```python
print('Predicted class:', np.argmax(output))
```
