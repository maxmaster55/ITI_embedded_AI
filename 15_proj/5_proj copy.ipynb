```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, regularizers
from tensorflow.keras.applications import ResNet50
import tensorflow_datasets as tfds
import pandas as pd
import os
from glob import glob
```

```python

data_files = glob(os.path.join("PAMAP2_Dataset/Protocol", "**", "*.dat"), recursive=True)

#load all into a single DataFrame
dfs = []
for f in data_files:
    df = pd.read_csv(f, sep=' ', header=None, engine='python')  # PAMAP2 uses space separator
    dfs.append(df)

data = pd.concat(dfs, ignore_index=True)
print("Shape of raw data:", data.shape)
```
