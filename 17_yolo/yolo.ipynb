```python
from ultralytics import YOLO
```

```python
DATA_YEML = "coco128.yaml"

MODEL_BASE = "yolov8n.pt"
```

```python
model = YOLO(MODEL_BASE)
```

```python
model.train(data=DATA_YEML, epochs=10, imgsz=640, batch=16, patience=3, project="runs/detect", name="yolo_coco128")
```

```python
best = YOLO("runs/detect/runs/detect/yolo_coco128/weights/best.pt")
```

```python
best.val()
```

```python
import cv2
import matplotlib.pyplot as plt
import pathlib
import os
```

```python
pred = best.predict(
    source="https://ultralytics.com/images/bus.jpg",
    conf=0.25,
    save=True
)
```

```python
save_dir=pathlib.Path(pred[0].save_dir)
```

```python
out_img = next(save_dir.glob("bus.jpg"))
```

```python
img = cv2.cvtColor(cv2.imread(str(out_img)), cv2.COLOR_BGR2RGB)
```

```python
plt.figure(figsize=(10, 10))
plt.imshow(img)
plt.axes("off")
plt.show()
```
